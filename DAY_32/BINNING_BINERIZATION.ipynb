Let's understand **Discretization** in Machine Learning.

### 1. What is Discretization?

Discretization is the process of converting a continuous numerical variable into discrete intervals (bins).

For example, suppose we have an Age column:

| Original Age | Discretized Age |
| ------------ | --------------- |
| 12           | 0–20            |
| 18           | 0–20            |
| 25           | 20–40           |
| 35           | 20–40           |
| 45           | 40–60           |
| 55           | 40–60           |

Instead of having individual numerical values, we divide them into groups.

### 2. Why do we use it?

* Simplifies continuous numerical data.
* Helps identify patterns in certain datasets.
* Can make some nonlinear relationships easier for models to capture.
* Useful for algorithms that work with categorical or interval-based features.

### 3. Types of Discretization

**A. Equal Width Discretization**

Every bin has the same width.

Example: Age from 0 to 100, divided into 5 bins.

Each bin has a width of 20.

**B. Equal Frequency Discretization**

Every bin contains approximately the same number of observations.

For example, 100 observations divided into 5 bins means approximately 20 observations per bin.

### 4. Implementation in Scikit-learn

We use `KBinsDiscretizer`.

```python
from sklearn.preprocessing import KBinsDiscretizer

dis = KBinsDiscretizer(
    n_bins=5,
    encode='ordinal',
    strategy='quantile'
)

X_transformed = dis.fit_transform(X)
```

Parameters:

* `n_bins=5`: Creates 5 bins.
* `encode='ordinal'`: Represents bins as 0, 1, 2, 3, 4.
* `strategy='quantile'`: Uses equal-frequency discretization.

Other strategies:

* `uniform`: Equal-width bins.
* `quantile`: Equal-frequency bins.
* `kmeans`: Uses K-Means clustering to create bins.

**Remember:** Discretization converts continuous numerical data into discrete intervals.

````python
import pandas as pd ; 
import numpy as np ; 
import matplotlib.pyplot as plt 

from sklearn.model_selection  import train_test_split 

from sklearn.compose import ColumnTransformer 

from sklearn.tree import DecisionTreeClassifier 

from sklearn.metrics import accuracy_score 

from sklearn.model_selection import cross_val_score 

from sklearn.preprocessing import KBinsDiscretizer

 
````

````python
df = pd.read_csv(r"E:\MACHINE_LEARNING\train.csv") 

df = df[ ['Age' ,'Fare' , "Survived"]]
````

````python
df.dropna(inplace = True )

df.head(1)

````

````python
X = df.iloc[:, :2]
y = df.iloc[:, 2]
````

````python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.2 , random_state = 34 )
````

````python
clf = DecisionTreeClassifier() 
clf.fit(X_train , y_train)

y_pred = clf.predict(X_test) 
````

````python
accuracy_score( y_test , y_pred )
````

````python
np.mean(cross_val_score( DecisionTreeClassifier() , X , y , cv = 15 , scoring = 'accuracy'))
````

````python
kbin_Age =  KBinsDiscretizer(n_bins = 15  , encode= 'ordinal' , strategy = 'quantile')
kbin_Fare = KBinsDiscretizer(n_bins = 15  , encode= 'ordinal' , strategy = 'quantile')
````

````python
trf = ColumnTransformer([('first' , kbin_Age , [0]) , ( 'second' , kbin_Fare , [1]) ] )
````

````python
X_train_trans = trf.fit_transform(X_train) 
X_test_trans  = trf.transform(X_test) 


````

````python
output = pd.DataFrame({
    'age':X_train['Age'],
    'age_trf':X_train_trans[:,0],
    'fare':X_train['Fare'],
    'fare_trf':X_train_trans[:,1]
})
````

````python
accuracy_score(y_test , y_pred_trans)
````

````python

def discretize(bins, strategy):

    kbin_Age = KBinsDiscretizer(
        n_bins=bins, encode='ordinal', strategy=strategy
    )

    kbin_Fare = KBinsDiscretizer(
        n_bins=bins, encode='ordinal', strategy=strategy
    )

    trf = ColumnTransformer([
        ('first', kbin_Age, [0]),
        ('second', kbin_Fare, [1])
    ])

    pipe = Pipeline([
        ('pre', trf),
        ('model', DecisionTreeClassifier())
    ])

    score = np.mean(cross_val_score(
        pipe, X, y, cv=14, scoring='accuracy'
    ))

    print("Accuracy:", score)

    # Transform data only for visualization
    X_trf = trf.fit_transform(X)

    plt.figure(figsize=(14, 4))

    plt.subplot(121)
    plt.hist(X['Age'])
    plt.title("Age Before")

    plt.subplot(122)
    plt.hist(X_trf[:, 0], color='red')
    plt.title("Age After")

    plt.show()

    plt.figure(figsize=(14, 4))

    plt.subplot(121)
    plt.hist(X['Fare'])
    plt.title("Fare Before")

    plt.subplot(122)
    plt.hist(X_trf[:, 1], color='red')
    plt.title("Fare After")

    plt.show()


````
