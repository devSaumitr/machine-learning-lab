# working with date 


```python
date['date'] = pd.to_datetime(date['date'])
date.info()

```

```python
# extract year 

date['date_year' ] = date['date'].dt.year
```

```python
date.head()
```

```python
# extract month

date['date_month' ] = date['date'].dt.month

date.head()
```

```python
# extract month

date['date_month' ] = date['date'].dt.month_name()

date.head()
```

```python
# extract day

date['date_day' ] = date['date'].dt.day

date.head()
```

```python
# extract month

date['week of the day' ] = date['date'].dt.dayofweek

date.head()
```

```python
# extract day

date['date_dayname' ] = date['date'].dt.day_name()

date.head()
```

```python
# is weeekwnd 


date['date_is_weekend'] = np.where(date['date_dayname'].isin(['Sunday' , 'Saturday']) , 1, 0 )


date.drop(columns= ['product_id', 'city_id' , 'orders'])


date.head()
```

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# extract day

date['quarter' ] = date['date'].dt.quarter

date.head()
```

```python
# semester 


date['semester'] = np.where(date['quarter'].isin([1 ,2 ]) , 1, 2  )


date.drop(columns= ['product_id', 'city_id' , 'orders'])


date.head()
```
