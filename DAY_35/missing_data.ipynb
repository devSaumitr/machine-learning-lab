```python
import numpy as np 
import pandas as pd 


import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv(r'E:\MACHINE_LEARNING\DAY_35\data_science_job.csv') 
```

```python
df.head(3)
```

```python
df.isnull().mean()*100 
```

```python
cols = [var for var in df.columns if df[var].isnull().mean() < 0.05 and df[var].isnull().mean() > 0 ] 

cols
```

```python
# check out for the loss occured 


len(df[cols].dropna()) / len(df)


```

```python

new_df = df[cols].dropna() 
df.shape , new_df.shape
```

```python
new_df.hist(bins= 50 , density= True , figsize=  ( 12 , 12 ) ) 
plt.show()
```

```python
fig = plt.figure(figsize =(12 ,8  ))
ax1 = fig.add_subplot(121)

# original data
df['training_hours'].hist(bins=50, ax=ax1, density=True, color='red')

# data after cca, the argument alpha makes the color transparent, so we can

# see the overlay of the 2 distributions
new_df['training_hours'].hist(bins=50, ax=ax1, color='green', density=True, alpha=0.8)




ax2 = fig.add_subplot(122)

# original data
df['training_hours'].plot.density(color='red')

# data after cca
new_df['training_hours'].plot.density(color='green')
```
