## Fetching JSON data from a public API

We request ready-made JSON from a web API and turn it into a table — no local file involved.


### 1. `requests.get` + `response.json()`
`requests.get(url)` returns an HTTP response; `.json()` parses its body into Python (list/dict). No local file involved — the data lives on the server.

````python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

data = response.json()

df = pd.DataFrame(data)

print(df)
````

### Example: one raw record from the movies API

Each record carries many fields (`movie_id`, `overview`, `popularity`, ...). The code below keeps only what we need:

```
original_title : "tick, tick... BOOM!"
popularity     : 21.04
```


### 2. JSON → DataFrame → column selection
`pd.DataFrame(result['data'])[['original_title','popularity']]` keeps only the needed columns. The stray movie-record markdown in the original is unrelated text, not code.

````python
import pandas as pd 
import requests as rq 

url = "https://jsonfakery.com/movies/simple-paginate"
rs = rq.get(url)
result = rs.json()


pd.DataFrame(result["data"])[["original_title","popularity"]]
````

## Key takeaways (Day 17)
- Pattern: GET → `.json()` → `DataFrame` → inspect. Reproduce with `random_state`-free live data (values change per call).
- **Gaps:** first API response is never displayed; no status-code check (`response.raise_for_status()`), pagination, or cleaning.
- **Common mistakes:** assuming the JSON key (`data`) without printing keys first; no handling of network failure/rate limits.
