## 1. Imports and Setup

Matplotlib provides low-level plotting control. Seaborn builds on Matplotlib and offers convenient statistical plots, especially for pandas DataFrames.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set_style("whitegrid")
```

## 2. Matplotlib Basics

### Line plot
**Use:** Show trends or changes across an ordered variable, such as time.

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 6]

plt.plot(x, y, color="red", linestyle="--", marker="o", label="Series")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot")
plt.grid(True)
plt.legend()
plt.show()
```

### Scatter plot
**Use:** Examine the relationship between two numerical variables.

```python
plt.scatter(x, y, s=50, alpha=0.7)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot")
plt.show()
```

### Bar chart
**Use:** Compare numerical values across categories. A bar chart is different from a histogram: bars compare categories, while histograms show a numerical distribution.

```python
categories = ["A", "B", "C"]
values = [12, 19, 8]

plt.bar(categories, values, width=0.6)
plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Bar Chart")
plt.show()

plt.barh(categories, values)
plt.title("Horizontal Bar Chart")
plt.show()
```

### Histogram
**Use:** Understand the distribution of a numerical feature. Look for skewness, unusual ranges, and possible outliers.

```python
ages = [18, 19, 20, 20, 21, 22, 22, 23, 24, 25, 27, 30, 35, 42]

plt.hist(ages, bins=8, edgecolor="black", alpha=0.7)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

### Box plot
**Use:** Summarize median, quartiles, spread, and potential outliers.

- Median: line inside the box
- Q1 and Q3: 25th and 75th percentiles
- IQR: Q3 − Q1
- Points beyond whiskers: potential outliers, not automatically errors

```python
salaries = [22000, 25000, 27000, 28000, 30000, 32000, 35000, 38000, 42000, 90000]

plt.boxplot(salaries)
plt.ylabel("Salary")
plt.title("Salary Box Plot")
plt.show()
```

### Pie chart
**Use:** Show proportions for a small number of categories. For many categories, a bar chart is often easier to compare.

```python
plt.pie(
    [45, 35, 20],
    labels=["A", "B", "C"],
    autopct="%1.1f%%",
    startangle=90,
    explode=[0, 0.05, 0]
)
plt.title("Category Proportions")
plt.show()
```

## 3. Figure, Axes, and Subplots

- **Figure:** The complete canvas.
- **Axes:** One plotting area.
- **Axis:** The x-axis or y-axis within an Axes object.

The object-oriented Matplotlib interface is useful when you need several plots or precise control.

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker="o")
ax.set_title("Object-Oriented Plot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

ax[0, 0].plot(x, y)
ax[0, 0].set_title("Line")

ax[0, 1].scatter(x, y)
ax[0, 1].set_title("Scatter")

ax[1, 0].hist(ages, bins=8)
ax[1, 0].set_title("Histogram")

ax[1, 1].boxplot(salaries)
ax[1, 1].set_title("Box Plot")

plt.tight_layout()
plt.show()
```

### Common customization commands

| Goal | Syntax |
|---|---|
| Figure size | `plt.figure(figsize=(10, 6))` |
| Title | `plt.title("Title")` |
| X label | `plt.xlabel("X")` |
| Y label | `plt.ylabel("Y")` |
| Legend | `plt.legend()` |
| Rotate ticks | `plt.xticks(rotation=45)` |
| Grid | `plt.grid(True)` |
| Transparency | `alpha=0.5` |
| Layout | `plt.tight_layout()` |
| Save | `plt.savefig("plot.png", dpi=300, bbox_inches="tight")` |
| Close | `plt.close()` |

Save a figure before `plt.show()` to avoid empty saved figures in some environments.

## 4. Seaborn Basics

For these examples, create a small sample DataFrame. You can replace it with your own dataset later.

```python
df = pd.DataFrame({
    "Age": [19, 22, 25, 28, 32, 35, 21, 27, 30, 40, 24, 33],
    "Salary": [22000, 28000, 35000, 42000, 52000, 60000, 26000, 39000, 48000, 75000, 32000, 58000],
    "Gender": ["F", "M", "F", "M", "F", "M", "M", "F", "M", "F", "F", "M"],
    "Purchased": ["No", "No", "Yes", "Yes", "Yes", "Yes", "No", "Yes", "Yes", "Yes", "No", "Yes"],
    "Year": [2019, 2020, 2021, 2022, 2023, 2024, 2019, 2020, 2021, 2022, 2023, 2024],
    "Sales": [100, 120, 150, 180, 210, 250, 90, 130, 160, 190, 220, 270]
})

df.head()
```

### Histogram — `histplot()`

Use for the distribution of a numerical feature. `kde=True` adds a smoothed density estimate.

```python
sns.histplot(data=df, x="Age", bins=8, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.histplot(data=df, x="Age", hue="Purchased", kde=True)
plt.title("Age Distribution by Purchase Class")
plt.show()
```

### KDE plot — `kdeplot()`

A KDE is a smoothed estimate of a distribution. Its shape depends on the smoothing bandwidth and can extend beyond observed values.

```python
sns.kdeplot(data=df, x="Salary", fill=True)
plt.title("Salary Density")
plt.show()
```

### Count plot — `countplot()`

Counts observations in each category. Use `hue` to compare category counts across another categorical variable.

```python
sns.countplot(data=df, x="Gender")
plt.title("Count by Gender")
plt.show()
```

```python
sns.countplot(data=df, x="Gender", hue="Purchased")
plt.title("Gender Counts by Purchase Class")
plt.show()
```

### Bar plot — `barplot()`

By default, estimates a statistic (usually the mean) for a numerical variable in each category and displays uncertainty intervals. You can pass a callable such as `np.median` to `estimator`.

```python
sns.barplot(data=df, x="Gender", y="Salary", estimator=np.median)
plt.title("Median Salary by Gender")
plt.show()
```

### Box plot — `boxplot()`

Compare numerical distributions across categories. `showfliers=False` hides points beyond the whiskers; it does not remove those observations from the dataset.

```python
sns.boxplot(data=df, x="Gender", y="Salary")
plt.title("Salary Distribution by Gender")
plt.show()
```

### Violin plot — `violinplot()`

Shows an estimated density shape for a numerical distribution across categories. It complements a box plot rather than replacing every detail of it.

```python
sns.violinplot(data=df, x="Gender", y="Salary")
plt.title("Salary Density by Gender")
plt.show()
```

### Scatter plot — `scatterplot()`

Use for relationships between two numerical features. `hue`, `style`, and `size` can encode additional variables.

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    hue="Purchased",
    style="Gender",
    size="Sales"
)
plt.title("Age vs Salary")
plt.show()
```

### Line plot — `lineplot()`

Useful for ordered data and trends. By default, repeated observations at an x-value may be aggregated and uncertainty intervals may be shown. Use `estimator=None` to draw individual observations rather than an aggregate estimate.

```python
sns.lineplot(data=df, x="Year", y="Sales", marker="o")
plt.title("Sales Trend")
plt.show()
```

### Regression plot — `regplot()`

Shows a scatter plot with a fitted regression relationship. This is useful for visual investigation, but it does not prove causation or guarantee predictive performance.

```python
sns.regplot(data=df, x="Age", y="Salary", ci=None)
plt.title("Age vs Salary with Regression Line")
plt.show()
```

## 5. Important EDA Plots

### Correlation heatmap

Correlation heatmaps are commonly used to inspect pairwise correlations between numerical features. Pearson correlation measures linear association.

```python
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.show()
```

### Pair plot

Shows pairwise relationships between numerical variables and distributions along the diagonal. It can become slow and crowded for large datasets, so select a few columns.

```python
sns.pairplot(
    df,
    vars=["Age", "Salary", "Sales"],
    hue="Purchased"
)
```

### Joint plot

Shows the relationship between two variables together with their individual distributions. Common `kind` values include `"scatter"`, `"kde"`, `"hist"`, `"hex"`, and `"reg"`.

```python
sns.jointplot(data=df, x="Age", y="Salary", kind="reg")
```

### General heatmap

A heatmap can display any two-dimensional numerical matrix, not just correlations. It can also visualize a confusion matrix.

```python
matrix = np.array([
    [10, 2, 1],
    [3, 12, 2],
    [1, 4, 9]
])

sns.heatmap(matrix, annot=True, cmap="viridis", fmt="d")
plt.title("Example Matrix Heatmap")
plt.show()
```

## 6. Which Plot Should You Choose?

| Question | Plot |
|---|---|
| How is one numerical feature distributed? | Histogram or KDE |
| How many observations are in each category? | Count plot |
| How do numerical values vary by category? | Box plot or violin plot |
| What is the mean or median by category? | Bar plot |
| Are two numerical features related? | Scatter plot |
| How does a value change over time? | Line plot |
| What is the fitted linear relationship? | Regression plot |
| Which numerical features are correlated? | Correlation heatmap |
| How do several numerical features relate? | Pair plot |
| What is the relationship plus each variable's distribution? | Joint plot |


## 7. Matplotlib vs Seaborn

| Feature | Matplotlib | Seaborn |
|---|---|---|
| Main purpose | Flexible plotting and detailed customization | Convenient statistical visualization |
| Line | `plt.plot(x, y)` | `sns.lineplot(data=df, x="Year", y="Sales")` |
| Scatter | `plt.scatter(x, y)` | `sns.scatterplot(data=df, x="Age", y="Salary")` |
| Histogram | `plt.hist(x)` | `sns.histplot(data=df, x="Age")` |
| Category counts | Usually count or prepare data first | `sns.countplot(data=df, x="Gender")` |
| Correlation heatmap | Build using Matplotlib primitives | `sns.heatmap(corr)` |
| Multiple plots | `plt.subplots()` | Often paired with Matplotlib subplots or Seaborn figure-level functions |

**Mental model:** Matplotlib gives you fine-grained control; Seaborn makes many statistical plots easier to create from DataFrames. Seaborn uses Matplotlib underneath.

### Important ML leakage rule

When visualization informs preprocessing or feature selection, make those decisions using the training data. Do not use the held-out test set to guide those decisions. Fit transformations on training data and apply the fitted transformations to the test data. A test set should represent an untouched final evaluation.


## 9. Common Mistakes

1. **Using a count plot for a continuous feature:** Prefer a histogram for a numerical distribution.
2. **Confusing bar plots and histograms:** Bar plots compare categories; histograms group numerical values into bins.
3. **Treating correlation as causation:** Correlation alone cannot establish cause and effect.
4. **Calling every box-plot outlier an error:** Points beyond whiskers may be valid observations.
5. **Forgetting aggregation:** Seaborn bar plots and line plots may aggregate repeated observations.
6. **Plotting too many variables at once:** Pair plots and heatmaps can become difficult to read.
7. **Ignoring missing values:** Missing data can affect plotting and interpretation.
8. **Using test data to make modeling decisions:** This can leak information into model development.
