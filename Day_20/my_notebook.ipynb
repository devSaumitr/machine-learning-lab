## Univariate analysis: one column at a time

Each plot below studies a single variable — its values and how often they occur.


```python
import pandas as pd 
import matplotlib.pyplot as plt    
import seaborn as sns 

# analysis  of the single colomn 


```

```python
df = pd.read_csv(r'E:\MACHINE_LEARNING\train.csv')
df.sample(5)
```

### Categorical columns

Count each level with bars (counts) or a pie (shares).


### 1. Categorical — `countplot`, `value_counts` pie
`countplot(x=df['Pclass'])` counts classes 1/2/3; `Survived.value_counts().plot.pie()` shows dead/alive share. Prefer `x='Pclass', data=df` form for cleaner labeling.

```python
sns.countplot(data = df , x =  df['Pclass'])   
```

```python
df["Survived"].value_counts().plot(kind = 'pie'  , autopct= "%.2f")
```

### Numerical columns

Study the distribution shape: center, spread, skew and outliers.


### Distribution shape with a histogram

Bars show how values spread across ranges — look for peaks, skew and gaps.


### 2. Numerical distribution — `hist`, `distplot`
`plt.hist(Age, bins=300)` is too fine (300 bins = noisy); `sns.distplot` is deprecated — use `histplot`/`kdeplot` in current seaborn. Read peak/spread/skew, not just the shape.

```python
plt.hist
```

```python
plt.hist(df["Age"] ,bins= 300)
```

```python
sns.distplot(df["Age"])
```

### Boxplot: spread and outliers

Box = middle 50% (Q1–Q3), line = median, whiskers = normal range, dots = outlier candidates (diagram attached).


### IQR fences for outliers

Lower fence = Q1 − 1.5·IQR, upper fence = Q3 + 1.5·IQR. Values beyond the fences are outlier candidates.


### Boxplots also show skew

A longer whisker on one side means a longer tail — here fares stretch far to the right.


### 3. Outliers — `boxplot(Fare)`, IQR rule
Box = Q1–Q3, line = median, whiskers = Q1−1.5·IQR / Q3+1.5·IQR; points beyond are candidate outliers (see attached `image.png`). `Age.min()` grounds the scale. Last cell crashed the kernel; `plt.hist` bare reference plots nothing by itself.

```python
sns.boxplot(data= df , x = df["Fare"])
```

```python
df['Age'].min()
```

## Key takeaways (Day 20)
- Univariate = one variable: counts/pie for categories; hist/dist/box for numbers.
- **Common mistakes:** 300 bins hides the shape; `distplot` no longer exists; `countplot(x=df['Pclass'])` should pass `data=`; bare `plt.hist` without call draws nothing.
