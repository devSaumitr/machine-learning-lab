### `ProfileReport` + `to_file`
`ProfileReport(df, title=..., explorative=True)` computes describe/missing/corr/interactions; `to_file` writes the self-contained HTML. Progress bar (`Describe Embarked`, `Summarize 47/47`, `Render HTML`, `Export`) is normal. Kernel crashed at 100% — the HTML may be incomplete; regenerate before trusting it.

```python
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv(r"E:\MACHINE_LEARNING\train.csv")

profile = ProfileReport(
    df,
    title="Titanic EDA Report",
    explorative=True
)

profile.to_file("titanic_report.html")
```
