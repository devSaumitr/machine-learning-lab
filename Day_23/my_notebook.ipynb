# Feature engineering: the job list

Raw tables are not model-ready. The jobs: **transformation** (change representation), **construction** (create new features), **selection** (keep what matters), **extraction** (build combined features).


## Missing values, types and scale

- sklearn needs numbers with no NaN: fill missing values first.
- Categorical → one-hot, strings → numbers; numbers → bands is called binning.
- Outliers strongly affect some algorithms.
- Scale features whose ranges differ hugely (e.g. Age vs Salary).


# Feature construction

Signals that do not exist as columns are created manually from what we have (e.g. family size from SibSp + Parch).


## Cleaning checklist

1. Fill or handle missing values.
2. Convert everything to numeric (encoding).
3. Treat outliers (select, then cap or remove).

Details for each step live in the [Day_24](../Day_24/)–[Day_29](../Day_29/) notebooks.


## Feature selection

Keep only features that actually help — fewer, better columns beat many noisy ones.


# Feature extraction

Build new combined features out of old ones (e.g. PCA in later study) instead of selecting a subset.
