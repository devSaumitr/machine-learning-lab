# Feature Scaling: Normalization in Machine Learning

## 1. What Is Normalization?

Normalization is a feature scaling technique used to bring numerical features into a specific range, usually between 0 and 1.

In a machine learning dataset, different features can have different scales.

For example:

| Age | Salary |
|---:|---:|
| 22 | 25000 |
| 25 | 40000 |
| 30 | 80000 |
| 35 | 120000 |

Here:
- Age ranges from 22 to 35.
- Salary ranges from 25,000 to 120,000.

Salary has much larger numerical values than age.

Some machine learning algorithms are sensitive to the scale of features. Therefore, we may need to scale these features before training the model.

Normalization helps transform the features into a comparable range.
