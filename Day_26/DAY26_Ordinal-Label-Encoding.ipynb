## Two kinds of data: numerical and categorical

Categorical splits further: **nominal** (equal status, no order — e.g. gender) vs **ordinal** (ranked levels — e.g. Poor < Average < Good). Target labels use label encoding.


### Rule: ordinal encoder for inputs, label encoder for the target

Ordered *input* columns get `OrdinalEncoder`; the *output* column gets `LabelEncoder`.


### Ordinal encoding maps ordered levels to numbers — you declare the order

Pass `categories=[["Poor","Average","Good"],...]` explicitly; never let the encoder sort alphabetically.

### Encoding the input columns

The next cells fit the ordinal encoder on the training split only.


```python
df =df.iloc[: , 2: ]


df
```

```python
from sklearn.model_selection import train_test_split as tts 

X_train , X_test , y_train ,y_test = tts(df.iloc[: , 0:2] , df.iloc[: , 2 ] ,test_size =0.3  , random_state = 34) 

 
```

```python
X_train.shape
```

### `OrdinalEncoder` with explicit order
`categories=[['Poor','Average','Good'],['School','UG','PG']]` fixes the order — alphabetical default would be wrong. `fit` on train, `transform` train (test likewise). Overwriting `X_train` with the array loses column names (preserved as-is; use DataFrame wrap in new code).

```python
from sklearn.preprocessing  import OrdinalEncoder as OE
```

```python
oe = OrdinalEncoder( categories= [['Poor' , 'Average' , "Good"] ,["School" ,"UG" , "PG"]])
```

```python
oe.fit(X_train)



```

```python
X_train =   oe.transform(X_train)
X_test = oe.transform(X_test) 
```

```python
X_train

oe.categories_
```

### Encoding the target column with `LabelEncoder`

`No`/`Yes` becomes `0`/`1` for the model output.


### `LabelEncoder` on the target
`fit_transform(y)` maps No→0/Yes→1. Only for targets, not inputs (no OneHot here despite the file title — OneHot is [Day 27](../Day_27/topic_notes.md)). No model trained — encoding only; 3 trailing code cells are empty.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le= LabelEncoder()

```

```python
le.fit(y_train)
```

```python
le.classes_
```

```python
y_train = le.transform(y_train)
y_test = le.transform(y_test )


```

## Key takeaways (Day 26)
- Ordered input → `OrdinalEncoder` (declare order!); target labels → `LabelEncoder`; unordered input → OneHot (next notebook).
- **Common mistakes:** letting the encoder sort categories alphabetically; encoding `y` with OneHot; fitting encoder on full data before split.
