```python
import pickle   

import numpy as np
```

```python
pipe  = pickle.load(open('pipe.pkl' ,'rb'))
```

## Step 1: create one raw passenger row

One raw 7-column Titanic row — the frozen pipeline handles all preprocessing itself.


```python
test_input  = np.array([3,	'male' ,	3.5	 ,0,	2	 ,18.2458 ,	'S'], dtype= object ).reshape(1,7)
```

### Why this is 1 line vs 11
`pipe` already contains imputer + encoders + scaler + selector + tree. `predict` runs them in order on the raw row. `UserWarning: X has no feature names but SimpleImputer fitted with names` x2 is harmless (numpy in, names out) — pass a DataFrame with column names to silence it.

```python
pipe.predict(test_input)
```
