```python
import pickle   

import numpy as np
```

```python


ohe_embarked = pickle.load(open(r'E:\MACHINE_LEARNING\BASICS\models\ohe_embarked.pkl' ,'rb'))
ohe_sex = pickle.load(open(r'E:\MACHINE_LEARNING\BASICS\models\ohe_sex.pkl' ,'rb'))
clf = pickle.load(open(r'E:\MACHINE_LEARNING\BASICS\models\clf.pkl' ,'rb'))

```

## Step 1: create one raw passenger row

The row below mimics a new passenger in raw Titanic format — encoding it by hand is the whole exercise.


```python
test_input  = np.array([3,	'female' ,	34.5	 ,0,	2	 ,15.2458 ,	'C'], dtype= object ).reshape(1,7)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:, 1].reshape(1, 1))
```

```python
test_input_embarked =ohe_embarked.transform(test_input[: ,-1].reshape(1,1))

test_input_embarked
```

```python
test_input_age = test_input[:,2].reshape(1 ,1)
```

### Manual reassembly
You must repeat the EXACT training order: numeric cols [0,3,4,5] + age + encoded sex + encoded embarked, same dtypes/shapes. Any deviation = silent wrong prediction or crash.

```python
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]],test_input_age,test_input_sex,test_input_embarked),axis=1)
```

```python
test_input_transformed.shape
```

```python
clf.predict(test_input_transformed)
```
