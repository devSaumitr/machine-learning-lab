## Why pipelines exist: chained preprocessing

Each step feeds the next, and repeating every step by hand at inference is error-prone. This notebook does the chain manually first — so you feel the pain pipelines remove.

### Impute → encode → concatenate (in the right order)
`SimpleImputer(strategy='mean'/'most_frequent')` fills Age/Embarked; `OneHotEncoder` expands Sex/Embarked; `np.concatenate` re-joins numeric + encoded blocks. The dropped-`drop` bug from the sibling notebook is FIXED here (assigned).

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer 
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier 
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv(r"E:\MACHINE_LEARNING\DATA_SETS\train.csv")

df.head()
```

```python
df = df.drop(columns= [	'PassengerId' ,'Name' , 'Ticket','Cabin'])
```

```python
X_train ,X_test , Y_train , Y_test = train_test_split(df.drop(columns= ['Survived'] )  , df['Survived'] ,test_size= .2 , random_state = 34) 
```

```python
X_train.sample(2)
```

```python
df.isnull().sum()
```

```python
si_age =SimpleImputer() 
si_embarked = SimpleImputer(strategy= 'most_frequent')
```

```python
X_train_age = si_age.fit_transform(X_train[['Age']])

X_train_embarked= si_embarked.fit_transform(X_train[['Embarked']])

X_test_age = si_age.transform(X_test[['Age']])

X_test_embarked= si_embarked.transform(X_test[['Embarked']])



```

```python
ohe_sex = OneHotEncoder(sparse_output = False, handle_unknown= 'ignore')

ohe_embarked = OneHotEncoder(sparse_output = False, handle_unknown= 'ignore')

# Handle igonre means if new value then make it zero 

X_train_sex = ohe_sex.fit_transform(X_train[['Sex']])


X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)



X_test_sex = ohe_sex.transform(X_test[['Sex']])


X_test_embarked = ohe_embarked.transform(X_test_embarked)
```

```python
X_train_rem =X_train.drop(columns= ['Sex' ,'Age' , 'Embarked' ])
X_test_rem =X_test.drop(columns= ['Sex' ,'Age' , 'Embarked' ])
```

```python
X_train_trans= np.concatenate((X_train_rem , X_train_age , X_train_sex , X_train_embarked) ,axis=1 )

X_test_trans= np.concatenate((X_test_rem , X_test_age ,X_test_sex , X_test_embarked) ,axis=1 )
```

```python
X_test_trans
```

```python

clf = DecisionTreeClassifier()


clf.fit(X_train_trans , Y_train)
```

```python
y_pred =clf.predict(X_test_trans)
```

```python
y_pred
```

### Train → evaluate → persist
`DecisionTree.fit(X_train_trans, y_train)`, `predict`, `accuracy_score` = 0.81005. `pickle.dump` saves the two encoders + classifier for the `Predict-Without-Pipeline` notebook.

```python
from sklearn.metrics import accuracy_score 


accuracy_score (Y_test , y_pred)
```

```python
import pickle 
```

```python
import os
import pickle

os.makedirs('models', exist_ok=True)

with open('models/ohe_embarked.pkl', 'wb') as f:
    pickle.dump(ohe_embarked, f)

with open('models/ohe_sex.pkl', 'wb') as f:
    pickle.dump(ohe_sex, f)

with open('models/clf.pkl', 'wb') as f:
    pickle.dump(clf, f)
```
